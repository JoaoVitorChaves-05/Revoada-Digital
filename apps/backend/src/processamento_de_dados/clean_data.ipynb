# Carregamento da tabela

```python
import pandas as pd 

df = pd.read_csv('Tabela_Escola_2025_V2.csv', sep=';', encoding='latin1')
view = df.head(10)
#print(view)

pd.set_option('display.max_columns', None)
for coluna in df.columns:
    print(coluna)

#NO_UF, NO_MUNICIPIO, CO_ENTIDADE, NO_ENTIDADE
```

# Limpeza da tabela

```python
colunas_desejadas = ['SG_UF', 'NO_MUNICIPIO', 'CO_ENTIDADE', 'NO_ENTIDADE']
colunas_ausentes = [coluna for coluna in colunas_desejadas if coluna not in df.columns]

if colunas_ausentes:
    raise KeyError(f'Colunas ausentes na tabela de origem: {colunas_ausentes}')

nova_tabela = df[colunas_desejadas].copy()
nova_tabela.to_csv('Tabela_Escola_2025_V3.csv', sep=';', index=False, encoding='UTF-8')

print(f'Tabela criada com {len(nova_tabela)} linhas e as colunas: {list(nova_tabela.columns)}')
```
